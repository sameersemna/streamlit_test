Deploying the Streamlit application via Colab
If we work on Google Colab, rather than downloading our files and creating a GitHub repo, we can deploy a Streamlit application directly via Colab using ngrok. ngrok is a proxy that lets you pass from a public URL to a private network (in our case, our local computer).

To use Streamlit via Google Colab with ngrok, we need to follow these steps:

```python
from pyngrok import ngrok
public_url = ngrok.connect(addr='8501')
public_url
```

Run the following code in a new Google Colab cell to create a .py file (called streamlit_app.py here). Then write the Python script associated with the Streamlit application in the cell.


```python
%%writefile streamlit_app.py
import streamlit as st
import pandas as pd

st.title('Sample Streamlit App')

st.write("Here's a sample Streamlit app displaying data from a CSV file.")

# Assume a file named 'sample_data.csv' exists in the Colab environment
try:
    df = pd.read_csv('sample_data.csv')
    st.write(df)
except FileNotFoundError:
    st.error("Error: sample_data.csv not found. Please upload the file or provide the correct path.")
```

Once the code is complete, run the following code in a new Google Colab cell to deploy the Streamlit application.

```python
!streamlit run /content/streamlit_app.py & npx localtunnel --port 8501
# password is the external URL IP address
```
